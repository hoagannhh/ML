```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# **DATASET**

```python
df = pd.read_csv('../data/pima.csv')
df.head()
```

```python
sns.countplot(x="Outcome", data=df, palette="Set2")
plt.title("Phân bố số lượng bệnh nhân tiểu đường và không tiểu đường")
plt.show()
```

```python
# Histogram cho từng thuộc tính
df.hist(bins=20, figsize=(15, 10))
plt.suptitle("Phân bố các thuộc tính", size=16)
plt.show()
```

```python
# Boxplot để so sánh theo Outcome
plt.figure(figsize=(15, 10))
for i, col in enumerate(df.columns[:-1], 1):
    plt.subplot(3, 3, i)
    sns.boxplot(x="Outcome", y=col, data=df, palette="Set3")
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
# Ma trận tương quan
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Ma trận tương quan giữa các thuộc tính")
plt.show()
```

```python
# barplot cho các thuộc tính chính
features = ["Glucose", "BMI", "Age", "Insulin", "BloodPressure"]

plt.figure(figsize=(15,8))
for i, col in enumerate(features, 1):
    plt.subplot(2, 3, i)
    sns.barplot(x="Outcome", y=col, data=df, palette="husl", ci="sd")
    plt.title(f"Trung bình {col} theo Outcome")

plt.tight_layout()
plt.show()
```
